```python
# Finding the percentage
if __name__ == '__main__':
    n = int(input())
    student_marks = {}
    for _ in range(n):
        name, *line = input().split()
        scores = list(map(float, line))
        student_marks[name] = scores
    query_name = input()
    print(f"{(sum(student_marks[query_name]) / len(student_marks[query_name])):.2f}")
```

```python
# List 
if __name__ == "__main__":
    n = int(input())
    numbers = []

    for _ in range(n):
        command, *args = input().split()

        if command == "insert":
            index, value = map(int, args)
            numbers.insert(index, value)
        elif command == "print":
            print(numbers)
        elif command == "remove":
            numbers.remove(int(args[0]))
        elif command == "append":
            numbers.append(int(args[0]))
        elif command == "sort":
            numbers.sort()
        elif command == "pop":
            numbers.pop()
        elif command == "reverse":
            numbers.reverse()
```

```python
# Nested Lists
dct = {}
if __name__ == '__main__':
    for _ in range(int(input())):
        name = input()
        score = float(input())
        dct[name] = score

set1 = set()
for i in (dct):
    set1.add(dct[i])

lst1 = list(set1)
lst1 = sorted(lst1)
for i in sorted(dct):
    if dct[i] == lst1[1]:
        print(i)
```
