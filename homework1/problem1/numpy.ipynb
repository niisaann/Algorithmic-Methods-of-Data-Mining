```python
# Arrays

import numpy

def arrays(arr):
    return numpy.array(arr, dtype=float)[::-1]

arr = input().strip().split(' ')
result = arrays(arr)
print(result)
```

```python
# Shape and Reshape

import numpy

numbers = numpy.array(input().split(), dtype=int)
matrix = numbers.reshape(3, 3)
print(matrix)
```

```python
# Transpose and Flatten

import numpy

n, _ = map(int, input().split())
rows = [input().split() for _ in range(n)]
matrix = numpy.array(rows, dtype=int)

print(matrix.T)
print(matrix.flatten())

```

```python
# Concatenate

import numpy

n, m, p = map(int, input().split())

first = numpy.array([input().split() for _ in range(n)], dtype=int)
second = numpy.array([input().split() for _ in range(m)], dtype=int)

first = first.reshape(n, p)
second = second.reshape(m, p)

print(numpy.concatenate((first, second), axis=0))

```

```python
# Zeros and Ones

import numpy

shape = tuple(map(int, input().split()))

print(numpy.zeros(shape, dtype=int))
print(numpy.ones(shape, dtype=int))

```

```python
# Eye and Identity

import numpy

numpy.set_printoptions(legacy="1.13")

n, m = map(int, input().split())
print(numpy.eye(n, m))

```

```python
# Array Mathematics

import numpy

n, _ = map(int, input().split())

a = numpy.array([input().split() for _ in range(n)], dtype=int)
b = numpy.array([input().split() for _ in range(n)], dtype=int)

print(a + b)
print(a - b)
print(a * b)
print(a // b)
print(a % b)
print(a ** b)

```

```python
# Floor, Ceil and Rint
import numpy
numpy.set_printoptions(legacy="1.13")

a = numpy.array(input().split(), dtype=float)

print(numpy.floor(a))
print(numpy.ceil(a))
print(numpy.rint(a))
```

```python
# Sum and Prod

import numpy

n, _ = map(int, input().split())
matrix = numpy.array([input().split() for _ in range(n)], dtype=int)

column_sums = numpy.sum(matrix, axis=0)
print(numpy.prod(column_sums))
```

```python
# Min and Max

import numpy

n, _ = map(int, input().split())
matrix = numpy.array([input().split() for _ in range(n)], dtype=int)

row_minimums = numpy.min(matrix, axis=1)
print(numpy.max(row_minimums))

```

```python
# Mean, Var, and Std

import numpy

n, _ = map(int, input().split())
matrix = numpy.array([input().split() for _ in range(n)], dtype=int)

print(numpy.mean(matrix, axis=1))
print(numpy.var(matrix, axis=0))
print(round(float(numpy.std(matrix)), 11))


```

```python
# Dot and Cross
 
import numpy

n = int(input())

a = numpy.array([input().split() for _ in range(n)], dtype=int)
b = numpy.array([input().split() for _ in range(n)], dtype=int)

print(numpy.dot(a, b))

```

```python
# Inner and Outer

import numpy
a = numpy.array(input().split(), dtype=int)
b = numpy.array(input().split(), dtype=int)

print(numpy.inner(a, b))
print(numpy.outer(a, b))
```

```python
# Linear Algebra

import numpy

n = int(input())
a = numpy.array([input().split() for _ in range(n)], dtype=float)

print(round(float(numpy.linalg.det(a)), 2))

```
