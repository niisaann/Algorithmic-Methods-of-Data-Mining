```python
# Introductions to sets
def average(array):
    unique_heights = set(array)
    return round(sum(unique_heights) / len(unique_heights), 3)
if __name__ == '__main__':
    n = int(input())
    arr = list(map(int, input().split()))
    result = average(arr)
    print(result)
```

```python
# Set .discard(), .remove() & .pop()

# Enter your code here. Read input from STDIN. Print output to STDOUT

input()  
numbers = set(map(int, input().split()))

command_count = int(input())
for _ in range(command_count):
    command, *args = input().split()

    if command == "pop":
        numbers.pop()
    elif command == "remove":
        numbers.remove(int(args[0]))
    elif command == "discard":
        numbers.discard(int(args[0]))

print(sum(numbers))
```

```python
# Set Mutations
# Enter your code here. Read input from STDIN. Print output to STDOUT
input() 
numbers = set(map(int, input().split()))

operation_count = int(input())
for _ in range(operation_count):
    operation, _ = input().split()
    other = set(map(int, input().split()))

    if operation == "update":
        numbers.update(other)
    elif operation == "intersection_update":
        numbers.intersection_update(other)
    elif operation == "difference_update":
        numbers.difference_update(other)
    elif operation == "symmetric_difference_update":
        numbers.symmetric_difference_update(other)

print(sum(numbers))
```

```python
# The Captain's Room
# Enter your code here. Read input from STDIN. Print output to STDOUT
k = int(input())
rooms = list(map(int, input().split()))

captain_room = (k * sum(set(rooms)) - sum(rooms)) // (k - 1)
print(captain_room)
```

```python
# Check Strict Superset
# Enter your code here. Read input from STDIN. Print output to STDOUT

a = set(map(int, input().split()))
n = int(input())

result = True
for _ in range(n):
    other = set(map(int, input().split()))
    if not (a > other):
        result = False
        break

print(result)
```
