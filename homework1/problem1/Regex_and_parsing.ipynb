```python
# Detect Floating Point Number
# Enter your code here. Read input from STDIN. Print output to STDOUT

import re

pattern = re.compile(r"[+-]?\d*\.\d+")

test_count = int(input())
for _ in range(test_count):
    value = input()
    print(pattern.fullmatch(value) is not None)
```

```python
# Group(), Groups() & Groupdict()
# Enter your code here. Read input from STDIN. Print output to STDOUT
import re

s = input()
match = re.search(r"([^\W_])\1", s)

print(match.group(1) if match else -1)
```

```python
# Re.findall() & Re.finditer()
# Enter your code here. Read input from STDIN. Print output to STDOUT

import re

s = input()
consonant = r"[b-df-hj-np-tv-z]"
pattern = rf"(?<={consonant})[aeiou]{{2,}}(?={consonant})"

matches = re.findall(pattern, s, flags=re.IGNORECASE | re.ASCII)

if matches:
    print(*matches, sep="\n")
else:
    print(-1)
```

```python
# Re.start() & Re.end()

# Enter your code here. Read input from STDIN. Print output to STDOUT
import re

s = input()
k = input()

pattern = rf"(?=({re.escape(k)}))"
matches = list(re.finditer(pattern, s))

if matches:
    for match in matches:
        print((match.start(1), match.end(1) - 1))
else:
    print((-1, -1))
```

```python
# Regex Substitution

# Enter your code here. Read input from STDIN. Print output to STDOUT

import re

pattern = r"(?<= )(?:&&|\|\|)(?= )"
replacements = {"&&": "and", "||": "or"}

n = int(input())
for _ in range(n):
    line = input()
    print(re.sub(pattern, lambda match: replacements[match.group()], line))
```

```python
# Validating phone numbers

# Enter your code here. Read input from STDIN. Print output to STDOUT

import re

pattern = re.compile(r"[789][0-9]{9}")

n = int(input())
for _ in range(n):
    number = input()
    print("YES" if pattern.fullmatch(number) else "NO")
```

```python
# Validating and Parsing Email Addresses

# Enter your code here. Read input from STDIN. Print output to STDOUT

import re

pattern = re.compile(r"<[A-Za-z][A-Za-z0-9._-]*@[A-Za-z]+\.[A-Za-z]{1,3}>")

n = int(input())
for _ in range(n):
    name, address = input().split(maxsplit=1)

    if pattern.fullmatch(address):
        print(name, address)
```

```python
# Hex Color Code

import re

n = int(input())
css = "\n".join(input() for _ in range(n))

color_pattern = r"#(?:[0-9A-Fa-f]{6}|[0-9A-Fa-f]{3})(?![\w-])"

for block in re.findall(r"\{([^{}]*)\}", css):
    for color in re.findall(color_pattern, block):
        print(color)
```

```python
# HTML Parser - Part 1

# Enter your code here. Read input from STDIN. Print output to STDOUT

from html.parser import HTMLParser


class TagParser(HTMLParser):
    def handle_starttag(self, tag, attrs):
        print(f"Start : {tag}")
        self.print_attributes(attrs)

    def handle_endtag(self, tag):
        print(f"End   : {tag}")

    def handle_startendtag(self, tag, attrs):
        print(f"Empty : {tag}")
        self.print_attributes(attrs)

    def print_attributes(self, attrs):
        for name, value in attrs:
            print(f"-> {name} > {value}")


n = int(input())
html = "\n".join(input() for _ in range(n))

parser = TagParser()
parser.feed(html)
parser.close()
```

```python
# HTML Parser - Part 2

from html.parser import HTMLParser

class MyHTMLParser(HTMLParser):
    def handle_comment(self, comment):
        if "\n" in comment:
            print(">>> Multi-line Comment")
        else:
            print(">>> Single-line Comment")
        print(comment)

    def handle_data(self, data):
        if data != "\n":
            print(">>> Data")
            print(data)
  
html = ""       
for i in range(int(input())):
    html += input().rstrip()
    html += '\n'
    
parser = MyHTMLParser()
parser.feed(html)
parser.close()
```

```python
# Detect HTML Tags, Attributes and Attribute Values

# Enter your code here. Read input from STDIN. Print output to STDOUT

from html.parser import HTMLParser


class TagParser(HTMLParser):
    def handle_starttag(self, tag, attrs):
        print(tag)
        for name, value in attrs:
            print(f"-> {name} > {value}")


n = int(input())
html = "\n".join(input() for _ in range(n))

parser = TagParser()
parser.feed(html)
parser.close()
```

```python
# Validating UID

# Enter your code here. Read input from STDIN. Print output to STDOUT

import re

pattern = re.compile(r"[A-Za-z0-9]{10}")

test_count = int(input())
for _ in range(test_count):
    uid = input()

    valid = (
        pattern.fullmatch(uid) is not None
        and len(set(uid)) == 10
        and sum(char.isupper() for char in uid) >= 2
        and sum(char.isdigit() for char in uid) >= 3
    )

    print("Valid" if valid else "Invalid")
```

```python
# Validating Credit Card Numbers

# Enter your code here. Read input from STDIN. Print output to STDOUT

import re

card_format = re.compile(r"[456][0-9]{3}(?:[0-9]{12}|(?:-[0-9]{4}){3})")
repeated_digit = re.compile(r"([0-9])\1{3}")

n = int(input())
for _ in range(n):
    card = input()
    valid = (
        card_format.fullmatch(card) is not None
        and repeated_digit.search(card.replace("-", "")) is None
    )
    print("Valid" if valid else "Invalid")
```
