```python
# Zipped!

# Enter your code here. Read input from STDIN. Print output to STDOUT

_, subject_count = map(int, input().split())

grades_by_subject = []
for _ in range(subject_count):
    grades_by_subject.append(list(map(float, input().split())))

for student_grades in zip(*grades_by_subject):
    average = sum(student_grades) / subject_count
    print(f"{average:.1f}")
    
```

```python
# ginortS

# Enter your code here. Read input from STDIN. Print output to STDOUT

def sort_key(char):
    if char.islower():
        return (0, char)
    if char.isupper():
        return (1, char)
    if char in "13579":
        return (2, char)
    return (3, char)


s = input()
print("".join(sorted(s, key=sort_key)))
```
